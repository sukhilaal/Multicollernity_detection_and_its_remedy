```python
from statsmodels.stats.outliers_influence import variance_inflation_factor
```

```python
import pandas as pd
from sklearn.datasets import fetch_california_housing

data = fetch_california_housing()
df = pd.DataFrame(data.data , columns=data.feature_names)
vif = pd.DataFrame()
vif["Features"] = df.columns
```

```python
for i in range(len(df.columns)):
    vif["VIF"] = [variance_inflation_factor(df.values , i ) for i in range(df.shape[1])]
```

```python
vif
```

```python
import numpy as np
import pandas as pd

# Initialize df1 as a copy of df to perform VIF-based feature selection
df1 = df.copy()

# Loop to drop features with VIF > 10
while True:
    features = df1.columns
    vif_data = pd.DataFrame()
    vif_data["Features"] = features

    # Calculate VIF for each feature
    # We need to ensure that there are at least two features to calculate VIF
    if len(features) < 2:
        print("Less than two features remaining. Cannot calculate VIF.")
        break
    try:
        vif_data["VIF"] = [variance_inflation_factor(df1.values, i) for i in range(len(features))]
    except np.linalg.LinAlgError as e:
        print(f"Caught a LinAlgError: {e}. This might happen if all features are perfectly correlated or if there's only one feature left after dropping. Breaking loop.")
        break

    # Get the maximum VIF value and its corresponding feature
    max_vif = vif_data["VIF"].max()
    # Check if max_vif is finite, if not, it means there might be issues (e.g. perfect multicollinearity leading to inf VIF)
    if not np.isfinite(max_vif):
        print("Encountered infinite VIF, indicating perfect multicollinearity. Review your features. Breaking loop.")
        # Drop the feature that caused the infinite VIF (e.g., first feature or a feature with a very high VIF if multiple are inf)
        feature_to_drop = vif_data.loc[vif_data['VIF'].isin([np.inf, -np.inf]), 'Features'].iloc[0]
        print(f"Attempting to drop '{feature_to_drop}' with infinite VIF.")
        df1 = df1.drop(columns=[feature_to_drop])
        continue


    feature_with_max_vif = vif_data.loc[vif_data["VIF"] == max_vif, "Features"].iloc[0]

    if max_vif > 10:
        print(f"Dropping feature '{feature_with_max_vif}' with VIF: {max_vif:.2f}")
        df1 = df1.drop(columns=[feature_with_max_vif])
    else:
        print("All VIFs are below 10. No more features to drop.")
        break

print("\nFinal DataFrame after VIF-based feature selection:")
print(df1.head())
print("\nFinal VIF values:")
print(vif_data)
```

```python
x = df1.iloc[:,:-1]# independent variable
y = df1.iloc[:,-1]# dependent variable
```
